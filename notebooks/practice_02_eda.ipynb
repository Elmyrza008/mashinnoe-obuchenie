## Выводы по графикам

1. Гистограмма показывает правостороннюю асимметрию: большинство счетов средние, но есть редкие большие значения.
2. Boxplot подтверждает наличие выбросов по `total_bill`.
3. Столбчатая диаграмма показывает, что заказы распределены по дням неравномерно.
4. Корреляция `total_bill` и `tip`: -0.053; это связь, а не доказательство причины.

Графики сохранены в папке `plots/`.


```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("../data/tips_dirty.csv")
display(df.info())
display(df.isna().sum())
display(df.describe(include="all"))

df["total_bill"].hist(bins=24)
plt.title("Распределение total_bill")
plt.xlabel("total_bill")
plt.ylabel("Частота")
plt.show()

df.boxplot(column="total_bill", vert=False)
plt.title("Boxplot total_bill")
plt.show()

df["day"].value_counts().plot(kind="bar")
plt.title("Частоты категорий day")
plt.xlabel("day")
plt.ylabel("Количество")
plt.show()
```
